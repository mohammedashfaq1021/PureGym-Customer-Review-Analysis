```python
import nltk
import matplotlib.pyplot as plt
from wordcloud import WordCloud
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from collections import Counter
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

## **Create a cleaning function**

```python
def clean_text(text):
  text = text.lower() # this is to bring the review to lowercase
  text = re.sub(r'[^a-z\s]'," ", text) #this keeps only letters between a to z. This means that punctuation and numbers will be removed.
  words = word_tokenize(text) # this will convert a sentence into words, also called tokenization.

  stop_words = set(stopwords.words('english'))
  words = [word for word in words if word not in stop_words] # this removes stopwords from my review

  return words
```

```python
google_df['cleaned'] = google_df['review'].apply(clean_text)
trustpilot_df['cleaned'] = trustpilot_df['review'].apply(clean_text)
```

# **Step 07: Falcon AI**

```python
!pip install transformers accelerate
```

```python
generator = pipeline('text-generation', model = 'tiiuae/falcon-7b-instruct', max_length = 512)
```

```python
sample_reviews = google_neg['review'].sample(5).tolist()
```

```python
def get_topic(review):
    prompt = f"""
    In the following customer review, pick out the three main topics.
    Return them as a numbered list.

    Review: {review}
    """

    result = generator(prompt)[0]["generated_text"]
    return str(result)
```

```python
topics_list = []

for review in sample_reviews:
    output = get_topic(review)
    topics_list.append(output)
    print(output)
    print("-" * 50)
```

```python
topics_list = [str(x) for x in topics_list if x is not None]
```

```python
print(topics_list)
```

```python
from bertopic import BERTopic
from umap import UMAP
from hdbscan import HDBSCAN
from sentence_transformers import SentenceTransformer

umap_model = UMAP(n_neighbors=2, n_components=2, min_dist=0.0, metric='cosine', random_state=42)
hdbscan_model = HDBSCAN(min_cluster_size=2, metric='euclidean', cluster_selection_epsilon=0.5, prediction_data=True)

# Initialize SentenceTransformer on CPU to avoid CUDA out of memory error
sentence_model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2", device="cpu")
topic_model_falcon = BERTopic(umap_model=umap_model, hdbscan_model=hdbscan_model, embedding_model=sentence_model)
topics_falcon, probs_falcon = topic_model_falcon.fit_transform(topics_list)
```

# **Step 08: Gensim LDA**

```python
!pip install gensim pyLDAvis
```

```python
import gensim
from gensim import corpora
import pyLDAvis
import pyLDAvis.gensim_models as gensimvis
```

```python
google_tokens = google_neg['cleaned'].tolist()
trust_tokens = trustpilot_neg['cleaned'].tolist()

all_negative_tokens = google_tokens + trust_tokens
```

```python
all_negative_tokens = [token for token in all_negative_tokens if len(token) > 0]
```

```python
dictionary = corpora.Dictionary(all_negative_tokens)
```

```python
corpus = [dictionary.doc2bow(text) for text in all_negative_tokens]
```

```python
lda_model = gensim.models.LdaModel(
    corpus = corpus,
    id2word = dictionary,
    num_topics = 10,
    random_state = 42,
    passes = 10
)
```

```python
topics = lda_model.print_topics(num_words = 10)

for topic in topics:
  print(topic)
  print("-" * 80)
```

```python
pyLDAvis.enable_notebook()

lda_visual = gensimvis.prepare(lda_model, corpus, dictionary)
lda_visual
```
